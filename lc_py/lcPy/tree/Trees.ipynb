````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

````python
from typing import List, Optional


class Solution:
        def sortedArrayToBST(self, nums: List[int]) -> Optional[TreeNode]:
            def helper(left, right):
                if left > right:
                    return None
                mid = (left + right) // 2
                node = TreeNode(nums[mid])
                node.left = helper(left, mid - 1)
                node.right = helper(mid + 1, right)
                return node
    
            return helper(0, len(nums) - 1)


nums = [-10, -3, 0, 5, 9]
s = Solution()
print(s.sortedArrayToBST(nums))
````

# 动态规划（Dynamic Programming）教程

## 1. 什么是动态规划？

动态规划是一种算法设计技术，用于解决具有**重叠子问题**和**最优子结构**性质的问题。

### 核心思想
- **记忆化**：将已经计算过的结果存储起来，避免重复计算
- **自底向上**：从最小的子问题开始，逐步构建更大问题的解
- **状态转移**：通过状态转移方程描述问题之间的关系

## 2. 动态规划的特征

### 重叠子问题（Overlapping Subproblems）
在递归求解过程中，同样的子问题会被多次计算。

### 最优子结构（Optimal Substructure）
问题的最优解包含子问题的最优解。

## 3. 动态规划解题步骤

1. **定义状态**：确定dp数组的含义
2. **状态转移方程**：找出状态之间的关系
3. **初始化**：确定边界条件
4. **遍历顺序**：确定计算顺序
5. **返回结果**：从dp数组中得到最终答案

## 4. 经典例题

### 例题1：斐波那契数列
```python
def fib(n):
    if n <= 1:
        return n
    
    # dp[i] 表示第i个斐波那契数
    dp = [0] * (n + 1)
    dp[0], dp[1] = 0, 1
    
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]  # 状态转移方程
    
    return dp[n]
```

### 例题2：爬楼梯问题
```python
def climbStairs(n):
    if n <= 2:
        return n
    
    # dp[i] 表示爬到第i阶楼梯的方法数
    dp = [0] * (n + 1)
    dp[1], dp[2] = 1, 2
    
    for i in range(3, n + 1):
        dp[i] = dp[i-1] + dp[i-2]  # 可以从i-1阶或i-2阶到达
    
    return dp[n]
```

### 例题3：最大子数组和（Kadane算法）
```python
def maxSubArray(nums):
    # dp[i] 表示以nums[i]结尾的最大子数组和
    dp = [0] * len(nums)
    dp[0] = nums[0]
    max_sum = nums[0]
    
    for i in range(1, len(nums)):
        # 要么接上前面的子数组，要么重新开始
        dp[i] = max(nums[i], dp[i-1] + nums[i])
        max_sum = max(max_sum, dp[i])
    
    return max_sum
```

## 5. 空间优化

很多DP问题可以进行空间优化，因为当前状态只依赖于前几个状态：

```python
# 斐波那契数列的空间优化版本
def fib_optimized(n):
    if n <= 1:
        return n
    
    prev2, prev1 = 0, 1
    for i in range(2, n + 1):
        current = prev1 + prev2
        prev2, prev1 = prev1, current
    
    return prev1
```

## 6. 常见DP类型

- **线性DP**：一维数组，如爬楼梯、最大子数组和
- **区间DP**：在区间上进行DP，如最长回文子串
- **背包DP**：0-1背包、完全背包等
- **树形DP**：在树上进行DP，如树的直径
- **状态机DP**：有限状态机，如股票买卖问题

# 找出有效子序列的最大长度


## 找出有效子序列的最大长度 - 动态规划解法

### 题目理解
给定一个整数数组 `nums` 和一个整数 `k`，找出最长的有效子序列。
有效子序列定义：对于子序列中相邻的两个元素，它们的和模 `k` 的结果相同。

### 动态规划思路分析

#### 1. 状态定义
`dp[i][r]` 表示：以第 `i` 个元素结尾，且相邻元素和模 `k` 余数为 `r` 的最长子序列长度

#### 2. 状态转移方程
对于当前元素 `nums[i]`，我们需要找到之前的元素 `nums[j]` (j < i)，
使得 `(nums[i] + nums[j]) % k == r`

转移方程：`dp[i][r] = max(dp[j][r] + 1)` for all valid j

#### 3. 优化思路
我们可以用 `max_len[r]` 来记录余数为 `r` 的最长子序列长度，
这样空间复杂度从 O(n*k) 优化到 O(k)


````python
from typing import List, Optional
class Solution:
    def maximumLength(self, nums: List[int], k: int) -> int:
        """
        动态规划解法
        时间复杂度: O(n * k)
        空间复杂度: O(k)
        """
        # max_len[r] 表示相邻元素和模k余数为r的最长子序列长度
        max_len = [0] * k
        
        for num in nums:
            # 当前元素的模k值
            curr_mod = num % k
            
            # 临时数组存储更新后的值，避免在遍历过程中修改原数组
            new_max_len = max_len[:]
            
            # 尝试所有可能的余数r
            for r in range(k):
                # 需要找到前一个元素prev，使得(curr + prev) % k == r
                # 即 prev % k == (r - curr_mod) % k
                prev_mod = (r - curr_mod) % k
                
                # 如果存在这样的前一个元素，更新当前余数r的最长长度
                if max_len[prev_mod] > 0 or prev_mod == curr_mod:
                    new_max_len[r] = max(new_max_len[r], max_len[prev_mod] + 1)
            
            max_len = new_max_len
        
        return max(max_len)

# 测试用例
solution = Solution()

# 示例1
nums1 = [1, 2, 3, 4, 5]
k1 = 2
print(f"示例1: nums={nums1}, k={k1}")
print(f"结果: {solution.maximumLength(nums1, k1)}")
print()

# 示例2
nums2 = [1, 4, 2, 3, 1, 4]
k2 = 3
print(f"示例2: nums={nums2}, k={k2}")
print(f"结果: {solution.maximumLength(nums2, k2)}")
````

### 算法详解

#### 核心思想
1. **状态压缩**：用一维数组 `max_len[r]` 记录每种余数对应的最长子序列长度
2. **状态转移**：对于当前元素，计算它与之前哪种余数的元素组合能产生目标余数
3. **动态更新**：遍历所有可能的目标余数，更新对应的最长长度

#### 关键点
- `prev_mod = (r - curr_mod) % k`：计算需要的前一个元素的余数
- 使用临时数组避免在遍历过程中修改原数组导致的问题
- 特殊处理：当 `prev_mod == curr_mod` 时，表示单个元素也可以形成长度为1的子序列

#### 时间复杂度分析
- 外层循环：O(n) - 遍历所有元素
- 内层循环：O(k) - 遍历所有可能的余数
- 总时间复杂度：O(n * k)

#### 空间复杂度
- 使用了长度为k的数组：O(k)
